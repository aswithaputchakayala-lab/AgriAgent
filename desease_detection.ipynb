```python
import tensorflow as tf
import numpy as np

IMG_SIZE = (224, 224)
BATCH_SIZE = 64
AUTOTUNE = tf.data.AUTOTUNE

def create_tf_dataset(indices, training=False):

    def generator():
        for i in indices:
            image = np.array(
                ds[i]["image"].convert("RGB"),
                dtype=np.float32
            )
            label = class_to_index[ds[i]["class_label"]]
            yield image, label

    dataset = tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(shape=(256, 256, 3), dtype=tf.float32),
            tf.TensorSpec(shape=(), dtype=tf.int32)
        )
    )

    dataset = dataset.map(
        lambda image, label: (
            tf.image.resize(image, IMG_SIZE) / 255.0,
            label
        ),
        num_parallel_calls=AUTOTUNE
    )

    if training:
        dataset = dataset.shuffle(
            5000,
            seed=42,
            reshuffle_each_iteration=True
        )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(AUTOTUNE)

    return dataset


train_ds = create_tf_dataset(train_idx, training=True)
val_ds = create_tf_dataset(val_idx, training=False)
test_ds = create_tf_dataset(test_indices, training=False)

print("✅ Fast pipeline created!")
```

```python
import tensorflow as tf
import numpy as np

IMG_SIZE = (160, 160)
BATCH_SIZE = 64
AUTOTUNE = tf.data.AUTOTUNE

def create_tf_dataset(indices, training=False):

    def generator():
        for i in indices:
            image = np.array(
                ds[i]["image"].convert("RGB"),
                dtype=np.float32
            )

            # Use the class index already provided by the dataset
            label = int(ds[i]["class_idx"])

            yield image, label

    dataset = tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(
                shape=(256, 256, 3),
                dtype=tf.float32
            ),
            tf.TensorSpec(
                shape=(),
                dtype=tf.int32
            )
        )
    )

    dataset = dataset.map(
        lambda image, label: (
            tf.image.resize(image, IMG_SIZE) / 255.0,
            label
        ),
        num_parallel_calls=AUTOTUNE
    )

    if training:
        dataset = dataset.shuffle(
            5000,
            seed=42,
            reshuffle_each_iteration=True
        )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(AUTOTUNE)

    return dataset


train_ds = create_tf_dataset(train_idx, training=True)
val_ds = create_tf_dataset(val_idx, training=False)
test_ds = create_tf_dataset(test_indices, training=False)

print("✅ DATA PIPELINE CREATED")
```

```python
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

train_labels = np.array([
    int(ds[i]["class_idx"])
    for i in train_idx
])

classes = np.arange(38)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_labels
)

class_weights = {
    int(i): float(w)
    for i, w in zip(classes, weights)
}

print("Number of class weights:", len(class_weights))
print("Minimum weight:", min(class_weights.values()))
print("Maximum weight:", max(class_weights.values()))
```

```python
import tensorflow as tf

IMG_SIZE = (160, 160)

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(160, 160, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers
base_model.trainable = False

# Input
inputs = tf.keras.Input(shape=(160, 160, 3))

# Augmentation
x = tf.keras.layers.RandomFlip("horizontal")(inputs)
x = tf.keras.layers.RandomRotation(0.10)(x)
x = tf.keras.layers.RandomZoom(0.10)(x)

# Convert our 0-1 images back to 0-255
x = x * 255.0

# MobileNetV2 preprocessing
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)

# Pretrained MobileNetV2
x = base_model(x, training=False)

# Classifier
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)

outputs = tf.keras.layers.Dense(
    38,
    activation="softmax"
)(x)

model = tf.keras.Model(inputs, outputs)

print("✅ Model created")
model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("✅ Model compiled")
```
